## 7.1.1 过滤缺失数据

```python
data = pd.Series([1,np.nan,3.5,np.nan,7])
data.dropna()
```

```python
data = pd.DataFrame([[1., 6.5, 3.], [1., np.nan, np.nan],
                     [np.nan, np.nan, np.nan], [np.nan, 6.5, 3.]])
data
```

```python
# dropna默认丢弃任何含有缺失值的行 返回新对象而不修改原始对象
data.dropna()
```

```python
# 传入how
data.dropna(how='all')
```

```python
data[4] = np.nan
data.dropna(axis=1,how='all')
```

```python
df = pd.DataFrame(np.random.randn(7,3))
df.iloc[:4,1] = np.nan
df.iloc[:2,2] = np.nan
df
```

```python
df.dropna()
```

```python
# thresh指定每行/列至少需要有多少个非空值才能保留
df.dropna(thresh=2)
```

## 7.1.2填充缺失数据

```python
df.fillna(0)
```

```python
# 若调用fillna时使用字典就可以实现对不同列填充不同的值
df = df.round(2)
df
```

```python
df.fillna({1:0.5,2:0})
```

```python
df = pd.DataFrame(np.random.randn(6,3))
df.iloc[2:,1] = np.nan
df.iloc[4:,2] = np.nan
df
```

```python
df.ffill()  # 前向填充
```

```python
# 限制只能填充2个
df.ffill(limit=2)
```

```python
# 横向填充
print(df.ffill(axis=1))
```
