## 3.2.2 返回多个值

```python
# 可以返回一个元组
def f():
    a=5
    b=10
    c=15
    return a,b,c  # 返回一个元组
x,y,z = f()  # 拆包元组
x
```

```python
# 也可以返回列表或字典
def g():
    a=5
    b=10
    c=15
    return {'a':a,'b':b,'c':c}  # 返回一个字典
```

## 3.2.3 函数即对象

```python
def apply_to_list(func, lst):
    return [func(x) for x in lst]
```

## 3.2.4 匿名(lambda)函数

```python
# 3.1有
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, numbers))
```

```python
# 3.1有
strings = ['foo','card','bar','aaaa','abab']
strings.sort(key=lambda x: len(set(x)))
strings
```

## 3.2.5 生成器

```python
# 生成器以延迟的方式返回多值序列,每返回一个值之后暂停,直到请求下一个值时继续,节省内存
def squares(n):
    for i in range(n):
        yield i**2  # yield语句返回一个值并暂停函数状态
gen = squares(10)
for x in gen:
    print(x, end=' ')
```

### 生成器表达式

```python
# 创建方式就是把列表表达式两端的方括号换成圆括号
gen_exp = (x**2 for x in range(10))
gen
```

```python
# 可作为函数参数
sum_of_squares = sum(x**2 for x in range(10))
sum_of_squares
```
