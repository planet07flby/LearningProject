## 3.2.2 返回多个值

```python
# 可以返回一个元组
def f():
    a=5
    b=10
    c=15
    return a,b,c  # 返回一个元组
x,y,z = f()  # 拆包元组
x
```

```python
# 也可以返回列表或字典
def g():
    a=5
    b=10
    c=15
    return {'a':a,'b':b,'c':c}  # 返回一个字典
```

## 3.2.3 函数即对象

```python
def apply_to_list(func, lst):
    return [func(x) for x in lst]
```

## 3.2.5 生成器

```python
# 生成器以延迟的方式返回多值序列,每返回一个值之后暂停,直到请求下一个值时继续,节省内存
def squares(n):
    for i in range(n):
        yield i**2  # yield语句返回一个值并暂停函数状态
gen = squares(10)
for x in gen:
    print(x, end=' ')
```

### 生成器表达式

```python
# 创建方式就是把列表表达式两端的方括号换成圆括号
gen_exp = (x**2 for x in range(10))
gen
```

```python
# 可作为函数参数
sum_of_squares = sum(x**2 for x in range(10))
sum_of_squares
```

### itertools模块

```python
import itertools
names = ['Alan','Adam','Wes','Will','Albert','Steven']
for letter,names in itertools.groupby(names,lambda x:x[0]):
    print(letter,list(names))
```

```python
from itertools import chain

# 1. 合并多个列表
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = [7, 8, 9]
result = list(chain(list1, list2, list3))
print("合并列表:", result)  # [1, 2, 3, 4, 5, 6, 7, 8, 9]

# 2. 合并不同类型可迭代对象
numbers = [1, 2, 3]
letters = ['a', 'b', 'c']
range_obj = range(4, 7)
result = list(chain(numbers, letters, range_obj))
print("混合类型:", result)  # [1, 2, 3, 'a', 'b', 'c', 4, 5, 6]

# 3. 展开嵌套列表
nested = [[1, 2], [3, 4], [5, 6]]
flat = list(chain(*nested))
print("展开嵌套:", flat)  # [1, 2, 3, 4, 5, 6]

# 4. 惰性计算（节省内存）
chained = chain([1, 2, 3], [4, 5, 6])
print("迭代器类型:", type(chained))  # <class 'itertools.chain'>
for i in chained:
    if i > 3:
        print(f"找到大于3的数: {i}")
        break
```

```python
from itertools import combinations
letters = ['A', 'B', 'C', 'D']
combs = list(combinations(letters, 2))
combs
```

### 3.2.6 错误和异常处理

```python
try:
    # 可能引发异常的代码
    print("尝试执行代码...")
    result = 10 / 2
except ZeroDivisionError:
    # 处理特定异常
    print("不能除以零！")
except ValueError as e:
    # 处理另一种异常
    print(f"值错误: {e}")
except Exception:
    # 处理其他所有异常
    print("发生了未知错误")
else:
    # 只有当 try 块没有异常时执行
    print(f"计算成功！结果是: {result}")
finally:
    # 无论是否发生异常，都会执行
    print("清理资源...")
```
