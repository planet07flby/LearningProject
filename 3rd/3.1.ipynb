```python
words = ['apple','bat','bar','atom','book']
by_letter = {}
for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)
by_letter
```

```python
# setdefault方法简化上述代码
by_letter = {}
for word in words:
    letter = word[0]
    by_letter.setdefault(letter, []).append(word)
by_letter
```

## 3.1.5 内置的序列函数

### enumerate函数

```python
# 基本用法：同时获取索引和值
fruits = ['apple', 'banana', 'cherry', 'date']

print("方法1：直接使用enumerate")
for i, fruit in enumerate(fruits):
    print(f"索引 {i}: {fruit}")

print("\n方法2：指定起始索引")
for i, fruit in enumerate(fruits, start=1):
    print(f"第{i}个水果: {fruit}")

print("\n方法3：转换为列表")
enum_list = list(enumerate(fruits, start=10))
print(f"从10开始编号: {enum_list}")

# 实际应用：查找元素位置
target = 'cherry'
for i, fruit in enumerate(fruits):
    if fruit == target:
        print(f"\n找到'{target}'，位置在索引 {i}")
        break
```

### sorted函数

```python
# 基本排序
numbers = [3, 1, 4, 1, 5, 9, 2]
print("升序:", sorted(numbers))
print("降序:", sorted(numbers, reverse=True))

# 字符串排序
words = ["banana", "apple", "cherry", "date"]
print("字母顺序:", sorted(words))

# 按长度排序
print("按长度:", sorted(words, key=len))

# 自定义排序规则
students = [("Alice", 25), ("Bob", 20), ("Charlie", 23)]
print("按年龄:", sorted(students, key=lambda x: x[1]))

# 原列表不变
print("原列表:", numbers)
print("排序后新列表:", sorted(numbers))
```

### zip函数

```python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]

# 使用 enumerate + zip
for i, (name, age) in enumerate(zip(names, ages)):
    print(f"{i}: {name} ({age}岁)")
```

## 3.1.6 列表、字典和集合推导式

```python
strings = ['a','as','bat','car','dove','python']
[x.upper() for x in strings if len(x)>2]
```

```python
# 列表 [expr for value in collection if condition]
# 字典 {key_expr:value_expr for value in collection if condition}
# 集合 {expr for value in collection if condition}
```

```python
unique_lengths = {len(x) for x in strings}
unique_lengths
```

```python
loc_mapping = {val:index for index, val in enumerate(strings)}
loc_mapping
```

```python
# map函数
# 1. 基本用法：对每个元素应用函数
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, numbers))
print("平方:", squares)  # [1, 4, 9, 16, 25]

# 2. 使用内置函数
str_nums = ["1", "2", "3"]
int_nums = list(map(int, str_nums))
print("转整数:", int_nums)  # [1, 2, 3]

# 3. 多序列操作
list1 = [1, 2, 3]
list2 = [4, 5, 6]
sums = list(map(lambda x, y: x + y, list1, list2))
print("对应相加:", sums)  # [5, 7, 9]

# 4. 字符串处理
words = ["hello", "world", "python"]
upper_words = list(map(str.upper, words))
print("大写:", upper_words)  # ['HELLO', 'WORLD', 'PYTHON']
```

```python
all_data = [['John','Emily','Michael','Sarah'],
            ['Jane','Daniel','Laura','Kevin'],
            ['Chris','Anna','Tom','Jessica']]
names_wanted = []
for data in all_data:
    enough_name = [(str(name)+' good') for name in data if name.count('a')>=2]
    names_wanted.extend(enough_name)
names_wanted
```

```python
some_tuples = [(1,2,3),(4,5,6),(7,8,9)]
flattened = [x for tup in some_tuples for x in tup]
flattened
```
