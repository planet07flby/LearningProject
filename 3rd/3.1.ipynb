```python
# dict接收二元组列表 zip合并成元组
d = dict(zip(['a','b','c'], [1,2,3]))
d
```

```python
# get方法访问字典值 若键不存在则返回默认值None或指定默认值
d.get('b',0)
```

```python
words = ['apple','bat','bar','atom','book']
by_letter = {}
for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)
by_letter
```

```python
# setdefault方法简化上述代码
by_letter = {}
for word in words:
    letter = word[0]
    by_letter.setdefault(letter, []).append(word)
by_letter
```

```python
# 字典键一般不可变 要用列表当作键可以用tuple()转化为元组
```

## 3.1.4 集合

```python
set([2,2,2,1,3,3,4,5,5])
```

```python
{2,2,2,1,3,3,4,5,5}
```

```python
a = {1,2,3,4,5}
b = {3,4,5,6,7,8}
# 并集
print(a.union(b))
print(a|b)
# 交集
print(a.intersection(b))
print(a&b)
```
