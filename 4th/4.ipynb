## 4.索引与切片

```python
# 一维
np.random.seed(20)
arr = np.random.randint(1,100,10)
print(arr)
print(arr[0])
print(arr[:])   #一行
print(arr[2:5])   #左闭右开 输出第2到第4个（从零开始）
print(arr[slice(2,10,2)])   #start end step
print(arr[ (arr>50) & (arr<90) ])
```

```python
# 二维
np.random.seed(20)
arr = np.random.randint(1,50,(4,8))
print(arr)
print(arr[1,3])
print(arr[1,:])   #第一行 :表示所有
print(arr[:,1])
print(arr[1,2:5])   # 2:5同上表示2-4
print(arr[arr>20])   #会返回一维数组
print(arr[2][arr[2]>30])   #第二行中大于50的
```

```python
np.random.seed(100)
arr = np.random.randint(1,20,8)
print(arr)
print(np.sum(arr))   # 和
print(np.mean(arr))   # 平均值
print(np.median(arr))   # 中位数
print(np.var(arr))   # 方差
print(np.std(arr))   # 标准差
print(np.max(arr),np.argmax(arr))   # 最大值和最大值索引
print(np.min(arr),np.argmin(arr))   # 最小值和最小值索引
print(np.percentile(arr,70))    # 分位数
print(np.percentile(arr,0))
print(np.cumsum(arr))   # 累积和
print(np.cumprod(arr))   # 累积积
```

## 7.基本比较函数

```python
print(np.greater([3,4,5,6,7],4))
print(np.less([3,4,5,6,7],4))
print(np.equal([3,4,5,6,7],4))
print(np.equal([2,3,4],[4,5,6]))
```

```python
print(np.logical_and([1,0],[1,1]))
print(np.logical_or([1,0],[1,1]))
print(np.logical_not([1,0]))
```

```python
print(np.any([0,0,0,0]))   # 检查是否至少有一个元素为true
print(np.all([1,1]))   # 检查是否全部元素为true
```

```python
# 自定义
# np.where(条件，符合条件，不符合条件))
arr = np.array([1,2,3,4,5])
print(np.where(arr<3,arr,0))
print(np.where(arr>2,1,0))
print(np.where(arr>2,"big","small"))
```

```python
np.random.seed(0)
score = np.random.randint(10,101,20)
print(score)
print(np.where(
    score<60,"不合格",np.where(
        score<80,"良好","优秀"
    )
))
```

```python
# np.select(条件数组,返回结果数组,default=)
print(np.select([score>80,(score>=60)&(score<=80),(score<60)],
                ["优秀","良好","不合格"],
                default="fail"))
```

## 8.基本排序函数

```python
print(np.sort(score))   # 不改变score
print(np.argsort(score))
print(score)
score.sort()   # 改变了score
print(score)
```

```python
print(np.unique(score))   # 去重加排序
```

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
print(np.concatenate([arr1,arr2]))   # 拼接
```

```python
print(np.split(score,4))
```
