# 4.7 示例：随机漫步

```python
import random
import matplotlib.pyplot as plt
position = 0
walk = [position]
nsteps = 1000
for i in range(nsteps):
    step = 1 if random.randint(0,1) else -1
    position += step
    walk.append(position)
plt.plot(walk[:100])
```

```python
import numpy as np
import matplotlib.pyplot as plt
nsteps = 1000
np.random.seed(20)
steps = np.random.randint(0,2,nsteps)
walk = steps.cumsum()
plt.plot(walk[:100])
```

```python
# 首次穿过10的索引
(walk>=10).argmax()
```

```python
# 一次模拟多个随机漫步
import numpy as np
nwalks = 500
nsteps = 100
k =10
steps = np.where(np.random.randint(0,2,(nwalks,nsteps))==0,-1,1)
walks = steps.cumsum(axis=1)
crossing_times = (np.abs(walks)<=2).argsort()[-k]
crossing_times.mean()
```
