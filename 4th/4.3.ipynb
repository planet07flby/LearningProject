```python
import numpy as np
arr = np.arange(10)
print(np.sqrt(arr))
print(np.exp(arr))
# np.maximum(x,y)
```

```python
arr = np.random.randn(8)*5
arr
```

```python
# modf函数
remainder, whole_part = np.modf(arr)  # 分离小数部分和整数部分
print(remainder)
print(whole_part)
```

```python
# ufunc 可以接受一个out可选参数,用于存储结果
out = np.zeros_like(arr)
np.add(arr, 1, out=out)
out
```
