# 4.6 线性代数

```python
import numpy as np
from numpy.linalg import inv
X = np.random.randn(5,5)*5
X
```

```python
np.round(X @ inv(X),2)  # 保留两位有效数字
```

```python
# 常用numpy.linalg模块中的函数
# matrix_rank() 矩阵的秩
# det() 矩阵的行列式
# inv() 矩阵的逆
# eig() 矩阵的特征值和特征向量
# svd() 奇异值分解SVD
# lstsq() 最小二乘解
# solve() 线性方程组的解
# qr() QR分解
# cholesky() Cholesky分解
# pinv() 伪逆矩阵
# trace() 矩阵的迹
```
