# 5.3 描述性统计的汇总和计算

```python
import pandas as pd
import numpy as np
df = pd.DataFrame([[1.4,np.nan],[7.1,-4.5],[np.nan,np.nan],[0.75,-1.3]],index=['a','b','c','d'],columns=['one','two'])
print(df)
```

```python
print(df.sum())
```

```python
df.sum(axis=1)
```

```python
df.sum(axis=0,skipna=False)
```

```python
df.mean(axis=1)
```

```python
# 最小值和最大值的索引
df.idxmax()
```

```python
df.idxmin()
```

```python
df.cumsum()
```

```python
df1 = pd.DataFrame([[1.4,2.4],[7.1,4.5],[8.9,7.3],[0.75,1.3]],index=['a','b','c','d'],columns=['one','two'])
df1.pct_change()
```

```python
df1['one'].corr(df1['two'])   # 相关系数
```

```python
df1['one'].cov(df1['two'])   # 协方差
```

```python
df1.corr()
```

```python
df.cov()
```

```python
# corrwith方法可以计算一个dataframe中列或行与另一个Series或DataFrame的列或行之间的相关系数
df1.corrwith(df['one'])
```
