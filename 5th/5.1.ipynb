# 5.1 pandas的数据结构介绍

```python
import numpy as np
import pandas as pd
obj = pd.Series([4, 7, -5, 3])
obj
```

```python
obj.array
```

```python
obj.index
```

```python
obj2 = pd.Series([4,7,-5,3],index = ['d','c','b','a'])
obj2
```

```python
obj2['a']
```

```python
obj2[['c','a','b']]
```

```python
# Series几乎同Numpy (如obj2[obj2>0] obj2*2等操作同Numpy语法)
# Series在算数运算中会自动对齐索引标签
# Series索引可以通过赋值方式就地修改
```

```python
obj2.to_dict()
```

```python
obj3 = pd.Series(obj2,index=['a','b','c','d'])
obj3
```

```python
obj3.name = 'number'
obj3.index.name = 'number.'
obj3
```

```python
# 可以通过赋值方式修改列
# 索引方式返回的列是深拷贝 可以用copy方法解决
```

```python
df = pd.DataFrame(
    {
        'id':[1,2,3,4,5],
        'name':['a','b','c','d','e'],
        'age':[10,20,30,40,50]
    },index=[1,2,3,4,5],columns=['id','age','name']   # columns可以调整类的顺序
)
del df['id']
df
```

```python
df.to_numpy()
```
