# 5.2 基本功能

```python
import pandas as pd
import numpy as np
obj = pd.Series([4.5,7.2,-5.3,3.6], index=['a','b','c','d'])
obj2 = obj.reindex(['d','c','b','a'])
obj2
```

```python
odj3 = pd.Series(['Blue','purple','yellow'],index=[0,2,4])
odj3.reindex(np.arange(6),method='ffill')  # 前向填充
```

```python
# 还可以用loc重建索引
frame = pd.DataFrame(np.arange(9).reshape((3,3)),index=['a','c','d'],columns=['Ohio','Texas','California'])
print(frame.loc[['a','c','d'],['California','Texas']])
```

```python
print(frame.drop(['d','a']))  # 行列都可以删
```

```python
# loc切片是包括后端的
# 数据选取用iloc和loc最好 可以同之前列表出现-1等等 (.iloc[-1])
# loc可以使用bool型数组
data = pd.DataFrame([[1,0,0,0],[1,5,6,7],[8,9,10,11],[1,13,14,15]],columns=['one','two','three','four'],index=['ohio','colorado','utah','new'])
print(data)
```

```python
data.loc[data['three']>3,'three'] = 6
print(data)
```

```python
# df1.add(df2,fill_value=0)   # df1和df2相加，缺失值用0填充
df1 = pd.DataFrame(np.random.rand(4,3),columns=list('bde'),index=['Utah','Ohio','Texas','Oregon'])
np.abs(df1)
```

```python
df1.apply(lambda x: x.max()-x.min())
```

```python
df1.apply(lambda x: x.max()-x.min(),axis=1)
```

```python
def f(x):
    return pd.Series([x.min(),x.max()],index=['min','max'])
df1.apply(f)
```

```python
df1.map(lambda x: round(x, 2))
```

```python
# 降序 ascending=False
obj = pd.Series([4,7,-3,2,np.nan])
obj.sort_values(na_position='first')
```
