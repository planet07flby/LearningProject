# 6.3 与Web API交互

```python
import pandas as pd
import requests
url = 'https://api.github.com/repos/pandas-dev/pandas/issues'
resp = requests.get(url)
resp.raise_for_status()
resp
```

```python
data = resp.json()
data[0]['title']
```

```python
# data中每个元素都是包含Github问题页上的所有数据的字典
issues = pd.DataFrame(data,columns=['number','title','labels','state'])
issues
```
