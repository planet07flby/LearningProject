```python
first,second,third = pieces
first+'::'+second+'::'+third
```

```python
# 更快更python
'::'.join(pieces)
```

## 7.4.2 正则表达式

```python
import re
text = 'foo    bar\t baz   \tqux'
re.split(r'\s+',text)
```

```python
# 拆分一个字符串,分隔符为数量不定的一组空白符(制表符、空格、换行符)
# 可以用re.compile()函数将模式字符串编译成一个正则表达式对象,以便重复使用
regex = re.compile(r'\s+')
regex.split(text)
```

```python
# 如果希望得到匹配正则表达式的所有模式,则可以使用findall方法
regex.findall(text)
```

```python
# findall返回字符串所有匹配项 search返回第一个匹配项 match只匹配字符串的起始位置
text = """Dave dave@google.com
Steve steve@gmail.com
Rob rob@gmail.com
Ryan ryan@yahoo.com"""
pattern = r"[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}"

# 不区分大小写
regex = re.compile(pattern, flags=re.IGNORECASE)
```

```python
regex.findall(text)
```

```python
m = regex.search(text)
m
```

```python
text[m.start():m.end()]
```

```python
print(regex.match(text))
```

```python
# sub方法将匹配到的模式替换为指定字符串并返回新的字符串
print(regex.sub('REDACTED',text))
```

```python
pattern = r"([A-Z0-9._%+-]+)@([A-Z0-9.-]+)\.([A-Z]{2,4})"
regex = re.compile(pattern, flags=re.IGNORECASE)
m = regex.match("wesm@bright.net")
m.groups()
```

```python
regex.findall(text)
```

```python
print(regex.sub(r"Username: \1, Domain: \2, Suffix: \3", text))
```

## 7.4.3 pandas的字符串函数

```python
import pandas as pd
import numpy as np
data = {"Dave": "dave@google.com", "Steve": "steve@gmail.com",
        "Rob": "rob@gmail.com", "Wes": np.nan}
data = pd.Series(data)
data
```

```python
data.isna()
```

```python
data.str.contains("gmail")
```

```python
pattern = r"([A-Z0-9._%+-]+)@([A-Z0-9.-]+)\.([A-Z]{2,4})"
data.str.findall(pattern, flags=re.IGNORECASE)
```

```python
matches = data.str.findall(pattern, flags=re.IGNORECASE).str[0]
matches
```

```python
matches.str.get(1)
```

```python
data.str[:5]
```

```python
data.str.extract(pattern, flags=re.IGNORECASE)
```
