## 9.1.1 图和子图

```python
fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax2 = fig.add_subplot(2,2,2)
ax3 = fig.add_subplot(2,2,3)
ax3.plot(np.random.randn(50).cumsum(),color='blue',linestyle='dashed')
ax1.hist(np.random.randn(100),bins=20,color='white')
ax2.scatter(np.arange(30),np.arange(30)+3*np.random.randn(30));
```

### 调整子图周围的间距

```python
fig,axes = plt.subplots(2,2,sharex=True,sharey=True)
for i in range(2):
    for j in range(2):
        axes[i,j].hist(np.random.randn(500),bins=50,color='white')
fig.subplots_adjust(wspace=0,hspace=0)
```

## 9.1.2 颜色、标记和线型

```python
fig = plt.figure()
ax = fig.add_subplot()
ax.plot(np.random.randn(30).cumsum(),color='white',linestyle='dashed',marker='o');
plt.show()
```

```python
# matplotlib的线型plot函数接受一组x,y坐标,还可以接收一个表示颜色和线型的参数 如: ax.plot(Dataset,linestyle='--',color='green')
fig = plt.figure()
ax = fig.add_subplot()
data = np.random.randn(30).cumsum()
ax.plot(data,color='white',linestyle='dashed',label='default');
ax.plot(data,color='white',linestyle='dashed',drawstyle='steps-post',label='steps-post');
ax.legend()
```

## 9.1.3 刻度、标签和图例

#### 设置标题、轴标签、刻度以及刻度标签

```python
fig, ax = plt.subplots()
ax.plot(np.random.randn(1000).cumsum());
```

```python
fig, ax = plt.subplots()
ax.plot(np.random.randn(1000).cumsum());
ticks = ax.set_xticks([0,250,500,750,1000])
labels = ax.set_xticklabels(['one','two','three','four','five'],rotation=30,fontsize=8)
ax.set_xlabel('stages')
ax.set_title('My First matplotlib plot')
```

#### 添加图例

```python
# 传入label参数创建图例 若要去除不传入即可
fig,ax = plt.subplots()
ax.plot(np.random.randn(1000).cumsum(),color='yellow',label='one')
ax.plot(np.random.randn(1000).cumsum(),color='skyblue',label='two')
ax.plot(np.random.randn(1000).cumsum(),color='red',label='three')
ax.legend()
```

## 9.1.4 注释和绘制子图

```python
from datetime import datetime
import pandas as pd
fig, ax = plt.subplots()
data = pd.read_csv('../examples/spx.csv', index_col=0, parse_dates=True)
spx = data['SPX']
spx.plot(ax=ax, color='blue')
crisis_data = [(datetime(2007,10,11), 'Peak of bull market'),
               (datetime(2008,3,12), 'Bear Stearns Fails'),
               (datetime(2008,9,15), 'Lehman Bankruptcy')]
for date, label in crisis_data:
    ax.annotate(label, xy=(date, spx.asof(date) + 75), xytext=(date, spx.asof(date) + 225),
                arrowprops=dict(facecolor='green', headwidth=4, width=2, headlength=4),
                horizontalalignment='left', verticalalignment='top')
ax.set_xlim(['1/1/2007', '1/1/2011'])  # 当前绘图范围
ax.set_ylim([600, 1800])
ax.set_title('Important dates in the 2008-2009 financial crisis');
```
