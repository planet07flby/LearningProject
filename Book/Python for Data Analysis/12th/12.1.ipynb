# 12.1 pandas与模型代码的接口

```python
import pandas as pd
data = pd.DataFrame({
    'x0':[1,2,3,4,5],
    'x1':[0.01,-0.01,0.25,-4.1,0.],
    'y':[-1.5,0.,3.6,1.3,-2.]
})
data
```

```python
data.to_numpy()
```

```python
df2 = pd.DataFrame(data.to_numpy(), columns=['one','two','three'])
df2
```

```python
model_col = ['one','two']
df2.loc[:,model_col].to_numpy()
```

```python
df2['category'] = pd.Categorical(['a','b','a','a','b'],categories=['a','b'])
df2
```

```python
dummies = pd.get_dummies(df2.category,prefix='category')
data_with_dummies = df2.drop('category',axis=1).join(dummies)
data_with_dummies
```
