```python
import pandas as pd
d = pd.Timestamp('2015-05-02 10:22')
print(d)
print(d.year)
print(d.month)
print(d.day)
print(d.hour)
print(d.minute)
print(d.second)
print(d.quarter)
# d.is_year_end  is_month_end等等 打is选即可
```

```python
print(d.day_name())   # 星期几
print(d.to_period('D'))   # 转换为天
print(d.to_period('Q'))   # 季度
print(d.to_period('M'))   # 月度
print(d.to_period('W'))   # 周
```

```python
# csv 日期转换
df = pd.read_csv('Dataset/weather.csv',parse_dates=['date'])   # date这列直接变为datetime变量类型
```

```python
# 日期数据作为索引
df.set_index('date',inplace=True)   # 在more里写了
print(df.loc['2013-01':'2013-8'])
```

```python
# 时间间隔
d1 = pd.Timestamp('2013-01-01')
d2 = pd.Timestamp('2013-02-01')
d3 = d2-d1
print(d3)
```

```python
df['delta'] = df['date'] - df['date'][0]
# 对间隔切片
df.set_index('delta',inplace=True)
print(df.loc['10 days':'20 days'])
```
