```python
import pandas as pd
import numpy as np
np.random.seed(42)
scores = pd.Series(np.random.randint(50,101,10),index=['学生'+str(i) for i in range(1,11)])
print(scores.mean())
print(scores.max())
print(scores.min())
print(len(scores[scores>scores.mean()]))
print(scores)
```

```python
temperatures = pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])
print((temperatures[temperatures>30].keys()))
print(temperatures.mean())
# a = temperatures*-1
# print(a.sort_values()*-1)
print(temperatures.sort_values(ascending=False))   # 降序排序 不执行
arr = np.arange(len(temperatures)-1)
for i in range(len(temperatures)-1):
    arr[i]=abs(temperatures.iloc[i+1]-temperatures.iloc[i])
print(np.argmax(arr)+1)
print(temperatures.diff().abs().argmax())   # 差值 无绝对值
# to list()可以转化成list输出
```

```python
date = pd.date_range('2023-01-01',periods=10)
prices = pd.Series([102.3,103.5,105.1,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=date)
print(prices)
# 收益率 后项除以前项-1
a = prices.pct_change()
print(a.idxmax())
print(a.idxmin())
```

```python
z = pd.date_range('2022-01-01',periods=12,freq='MS')
sales = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=z)
sales
```

```python
# 季度平均
sales.resample('QS').mean()
```

```python
a = sales.pct_change()
```

```python
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0,100,24),index=pd.date_range('2025-01-01',periods=24,freq='h'))
# hourly_sales.resample('D').sum()
print(hourly_sales.between_time('8:00','22:00').sum()/hourly_sales.sum())
print(hourly_sales[(hourly_sales.index.hour>=8) & (hourly_sales.index.hour<=22)].sum()/hourly_sales.sum())
hourly_sales.drop(hourly_sales.between_time('8:00','22:00').index)
print(hourly_sales.nlargest(3))
```
