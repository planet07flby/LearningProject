# Series

## 1.Series创建

```python
import pandas as pd
import numpy as np
s = pd.Series([1,2,3,4,5])
print(s)
s = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(s)
s = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'],name='value')
print(s)
```

```python
s1 = pd.Series(s,index=['a','c','e'])
print(s1)
```

## 2.Series属性

```python
print(s.index)
print(s.values)
print(s.shape)
print(s.ndim)
print(s.dtype)
print(s.name)
print(s.loc['a':'c'])   # 显式索引 支持切片
print(s.iloc[1])   # 隐式索引
```

```python
print(s['a'])
print(s[s<3])
s['f']=6
print(s)
print(s.head(2))
print(s.tail(2))
```

# pandas dataframe

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])
df1 = pd.DataFrame({'第一列':s1,'第二列':s2})
print(df1)
df = pd.DataFrame(
    {
        'id':[1,2,3,4,5],
        'name':['a','b','c','d','e'],
        'age':[10,20,30,40,50]
    },index=[1,2,3,4,5],columns=['id','age','name']   # columns可以调整类的顺序
)
print(df)
```

```python
print(df.index)   # 行索引
print(df.columns)   # 列标签
print(df.values)
# print(df.ndim)
# print(df.dtypes)
print(df.shape)
print(df.size)
```

```python
print(df.T)
```

```python
print(df.loc[4])
print(df.iloc[3])
```

```python
print(df.loc[2:,'name'])
print(df.iloc[:,0])
```

```python
print(df.at[3,'age'])
print(df.iat[0,1])   # loc,iloc同at,iat
```

```python
print(df['name'])   # 获取name这一列 同at,loc
print(df.name)
```

```python
# 多列数据访问
print(df[['name','age']])
```

```python
df[(df.age>20) & (df.id<5)]
```

```python
# 随机抽样
df.sample(2)
```

```python
print(df.head(2))
print(df.tail(2))
```

```python
df
```

```python
print(df.isin(['f',20]))
```

```python
print(df.isna())
```

```python
print(df.age.sum())
print(df.age.describe())
print(df.age.mean())
print(df.age.median())
print(df.age.quantile(0.25))
print(df.age.mode())
print(df.age.max())
print(df.age.min())
print(df.age.var())
print(df.count())   # 每列非缺失值的元素个数
print(df.value_counts())
df.drop_duplicates()
df.duplicated()   # 返回每行是否重复
df.duplicated(subset=['age'])   # 返回列是否重复
df.replace(20,30)   # 所有20变30
```

```python
df.cummax()
df.cummin()   # 想改变方向可以在括号里加axis=1 保证行方向不能有字符串
```

```python
df.sort_index()
df.sort_values(by=['age','id'],ascending=[False,True])   # 想降序可以加ascending=False  字典序
df.nlargest(2,columns=['age','id'])
# nsmallest同理
```
