```python
# 缺失值处理
s = pd.Series([1,2,np.nan,None,pd.NA])
print(s.isna())
```

```python
print(df.dropna())   # 剔除缺失值  注意对于dataframe有na的一行会被全删
```

```python
print(df.dropna(how='all'))   # 如果一行所有的值都是缺失值才会删
```

```python
df.dropna(thresh=2)   # 如果有n个不是缺失值就会删除否则保留
```

```python
df.dropna(axis=1)   # 删一整列
```

```python
df.dropna(subset=['age'])   # 如果某列有缺失值，则删除这一行
```

```python
# 填充
df.fillna({'temp_max':20,'wind':2.5})
df.fillna(df[['wind']].mean())
df.ffill()   # na值复制前一项的值
df.bfill()   # na值复制后一项的值
```

```python
# 去重
df.duplicated()   # 存在一整条重复返回true
df.drop_duplicates()   # 去重
df.drop_duplicates(subset=['name'])   # 根据指定列去重
df.drop_duplicates(subset=['name'],keep='last')   # 取最后的
```

```python
# 数据类型转换
df['name'] = df['name'].astype('float32')
df['gender'] = df['gender'].astype('category')
df['gender'] = df['gender'].map({'female':True,'male':False})
```

```python
pd.melt(df,id_vars=[],var_name='',value_name='')   # 宽表转长表 id_vars中是不变的
pd.pivot(df,index=[''],columns='',values='')        # 长表转宽表
df[['first','last']] = df['name'].str.split(' ',expand=True)   # 分裂
df.info()
```

```python
# 分箱
pd.cut(df['salary'],bins=2)   # 分区间 bins表示均分分成几个区间
pd.cut(df['salary'],bins=2).value_counts()   # 统计每个区间多少人
pd.cut(df['salary'],bins=[0,10000,20000,30000])   # bins=list  分成0-10000，10000-20000，20000-30000，30000-
df['range'] = pd.cut(df['salary'],bins=[0,10000,20000,30000],labels=['低','中','高'])   # bins=list  分成0-10000，10000-20000，20000-30000，30000-
pd.qcut(df['salary'],3)   # 人数等分  可以用value_counts方法验证
```

```python
df.set_index('name',inplace=True)   # inplace=True会改变使得name这一列变成索引
df.reset_index(inplace=True)
df.rename(columns={'age':'年龄'},index={0:4})   # 列的age改成年龄 0行该成4
df.index=[1,2,3,4]
df.columns=['','','']   # 批量修改
```

```python
# 分组聚合
df.groupby('department_id').groups   # 查看分组
df.groupby('department_id').get_group(20)   # 查看具体的分组数据
df1 = df.groupby('department_id')[['salary']].mean()   # 可以写多列
df1['salary'] = df1['salary'].round(2)  # 只留两位小数
```
