```python
import matplotlib.pyplot as plt
import random
# 折线图
# 创建图表，设置大小
plt.figure(figsize=(10,5))
month = ['1','2','3','4']
sales = [100,150,80,130]
# 绘制
plt.plot(month,sales,label='product A',marker='o')   # marker变加粗点
# 标题
plt.title('2025',color='skyblue',fontsize=20)
# 坐标标签
plt.xlabel('Month',color='orange')
plt.ylabel('Sales',color='orange')
# 图例（在绘制的label里写）
plt.legend(loc='upper left')
# 添加网格线
# plt.grid(True)
# plt.grid(axis='x')
plt.grid(axis='y',alpha=0.3,linestyle='--')   # alpha透明度  '--'虚线 '-'实线
# 设置刻度字体大小旋转
plt.xticks(rotation=0,fontsize=10)
# 设置y轴范围
plt.ylim(40,180)
# 数据点显示数值
for x,y in zip(month,sales):   # 相当于双for循环
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=8)   # 水平垂直
# 显示图表
plt.show()
```

```python
# 柱/条形图
# 创建图表，设置大小
plt.figure(figsize=(10,5))
subjects = ['maths','english','physics','chemistry']
scores = [85,92,78,88]
# 绘制
plt.bar(subjects,scores,label='david',width=0.6)   # width表示柱子宽度默认1   # barh会横过来放
# 标题
plt.title('2025',color='orange',fontsize=18)
# 图例（在绘制的label里写）
plt.legend(loc='upper left')
# 设置y轴范围
plt.ylim(0,100)
# 数据点显示数值
for x,y in zip(subjects,scores):   # 相当于双for循环
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=8)   # 水平垂直
# 坐标标签
plt.xlabel('Scores',color='pink')
plt.ylabel('Subjects',color='pink')
# 网格线
plt.grid(axis='y',alpha=0.3,linestyle='--')
# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 饼图
# 创建图表，设置大小
plt.figure(figsize=(10,8))
month = ['1','2','3','4']
sales = [100,150,80,130]
colors = ['#66b3ff','#99ff99','#ffcc99','#ff4499']   # 配色方案
# 绘制
plt.pie(sales,labels=month,autopct='%1.1f%%',shadow=False,startangle=90,colors=colors)
# 标题
plt.title('2025ratio',color='skyblue',fontsize=20)
# 自动优化排版
plt.tight_layout()
# 图例（在绘制的label里写）
plt.legend(loc='upper left')
# 显示图表
plt.show()
```

```python
# 环形图
# 创建图表，设置大小
plt.figure(figsize=(10,8))
month = ['1','2','3','4']
sales = [100,150,80,130]
colors = ['#66b3ff','#99ff99','#ffcc99','#ff4499']   # 配色方案
# 绘制
plt.pie(sales,labels=month,autopct='%1.1f%%',shadow=False,startangle=90,colors=colors,wedgeprops={'width':0.6})   # 圆环宽度
# 标题
plt.title('2025ratio',color='skyblue',fontsize=20)
# 自动优化排版
plt.tight_layout()
# 图例（在绘制的label里写）
plt.legend(loc='upper left')
# 显示图表
plt.show()
```

```python
# 爆炸式饼图
# 创建图表，设置大小
plt.figure(figsize=(10, 8))
month = ['1', '2', '3', '4']
sales = [100, 150, 80, 130]
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff4499']  # 配色方案
explode = [0.1,0,0,0]   # 设置突出块配置
# 绘制
plt.pie(sales, labels=month, autopct='%1.1f%%', shadow=True, startangle=90, colors=colors, explode=explode)
# 标题
plt.title('2025ratio', color='skyblue', fontsize=20)
# 自动优化排版
plt.tight_layout()
# 图例（在绘制的label里写）
plt.legend(loc='upper left')
# 显示图表
plt.show()
```

```python
# 散点图
# 创建图表，设置大小
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.figure(figsize=(10, 8),facecolor='white',label='random')
x = []
y = []
for i in range(1000):
    tmp = random.uniform(0,10)
    x.append(tmp)
    tmp2 = 2*tmp + random.gauss(0,2)
    y.append(tmp2)
# 绘制
plt.scatter(x,y,label='random',alpha=0.9,color='darkblue')
# 标题
plt.title('2025', color='skyblue', fontsize=20)
# 添加网格线
plt.grid(True,alpha=0.5,color='grey',linestyle='--')
# 图例（在绘制的label里写）
plt.legend(loc='upper left')
# 设置y轴范围
plt.ylim(0,30)
plt.xlim(0,10)
# 刻度
plt.xticks(rotation=0,fontsize=12,color='black')
plt.yticks(rotation=0,fontsize=12,color='black')
# 坐标标签
plt.xlabel('x',color='red')
plt.ylabel('y',color='red')
# 自动优化排版
plt.tight_layout()
# 坐标变黑
ax = plt.gca()
ax.spines['bottom'].set_color('black')  # x轴
ax.spines['left'].set_color('black')    # y轴
# 保存显示图表
plt.savefig('2026.png')
plt.show()
```
