# 11.2 时间序列基础知识

```python
import pandas as pd
import numpy as np
from datetime import datetime
dates = [datetime(2011,1,2), datetime(2011,1,5), datetime(2011,1,7), datetime(2011,1,8), datetime(2011,1,10), datetime(2011,1,12)]
ts = pd.Series(np.random.randn(6), index=dates)
ts
```

```python
ts + ts[::2]
```

```python
stamp = ts.index[0]
stamp
```

## 11.2.1 索引、选取、子集构造

####  可以传入字符串日期,datetime,timestamp来切片 与numpy切片运算相同 对切片修改会修改原数据

```python
stamp = ts.index[2]
ts[stamp]
```

```python
ts['20110112']
```

```python
# 对于较长的时间序列 只需传入年或年月即可选取切片
longer_ts = pd.Series(np.random.randn(1000), index=pd.date_range('2000-01-01', periods=1000))
longer_ts.head()
```

```python
longer_ts['2001']
```

```python
longer_ts['2001-05']
```

```python
ts[datetime(2011,1,7):]
```

```python
ts[datetime(2011,1,7):datetime(2011,1,10)]
```

```python
# 大部分时间序列数据都是按时间先后排序的 因此也可以用不存在于该时间序列中的时间戳对其进行范围查询
ts
```

```python
ts['20110106':'20110111']
```

```python
# 对dataframe也有效
dates = pd.date_range('2000-01-01', periods=100, freq='W-WED')
long_df = pd.DataFrame(np.random.randn(100, 4), index=dates, columns=['Colorado', 'Texas', 'New York', 'Ohio'])
long_df.loc['2001-05']
```

## 11.2.2 带有重复索引的时间序列

```python
dates = pd.DatetimeIndex(['2011-01-01', '2011-01-02', '2011-01-02', '2011-01-02', '2011-01-03'])
dup_ts = pd.Series(np.arange(5), index=dates)
dup_ts
```

```python
dup_ts.index.is_unique
```

```python
dup_ts['2011-01-03']
```

```python
dup_ts['2011-01-02']
```

```python
grouped = dup_ts.groupby(level=0).mean()
grouped
```
