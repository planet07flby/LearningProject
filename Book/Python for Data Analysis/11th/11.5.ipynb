```python
import pandas as pd
p = pd.Period('2011', freq='Y-DEC')  # Y-DEC表示以12月为年末 频率-锚定
p
```

```python
p + 5
```

```python
pd.Period('2014', freq='Y-DEC') - p
```

```python
periods = pd.period_range('2000-01-01', '2000-06-30', freq='M')
periods
```

```python
values = ['2001Q3', '2002Q2', '2003Q1']
index = pd.PeriodIndex(values, freq='Q-DEC')
index
```

## 11.5.1 周期的频率转换

```python
p = pd.Period('2011', freq='Y-DEC')
p
```

```python
p.asfreq('M', how='start')  # 转换为月频率 以月初为锚点
```

```python
p.asfreq('M', how='end')  # 以月末为锚点
```

```python
import numpy as np
periods = pd.period_range('2006', '2009', freq='Y-DEC')
ts = pd.Series(np.random.randn(len(periods)), index=periods)
ts
```

```python
ts.asfreq('M', how='start')
```

## 11.5.2 时间戳和周期的相互转换

```python
dates = pd.date_range('2000-01-01', periods=3, freq='MS')
ts = pd.Series(np.random.randn(len(dates)), index=dates)
ts
```

```python
pts = ts.to_period()
pts
```

```python
# 允许出现重复周期
dates = pd.date_range('2000-01-29', periods=6)
ts2 = pd.Series(np.random.randn(len(dates)), index=dates)
ts2.to_period('M')
```

```python
pts = ts2.to_period()
pts
```

```python
pts.to_timestamp()
```
