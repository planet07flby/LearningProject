```python
pd.date_range('2012-05-02 12:56:31', periods=5, normalize=True)
```

## 11.3.3 对超前或滞后数据进行移位

```python
ts = pd.Series(np.random.randn(4), index=pd.date_range('2000-01-01', periods=4, freq='MS'))  # MS月初
ts
```

```python
# 移位使得在时间序列前面和后面产生缺失数据
ts.shift(2)
```

```python
ts.shift(-2)
```

```python
# shift通常可以计算一个或多个时间序列中的连续百分比变化
ts / ts.shift(1) - 1
```

```python
# shift还可以接受一个freq参数,它会直接将数据移动到新的时间戳而不是简单地移动数据
ts.shift(2, freq='MS')
```

```python
# 还可以传入其他频率
ts.shift(3, freq='D')
```

```python
# 可以注意到shift没有修改原数组
```

### 通过偏移量对日期进行移位

```python
from datetime import datetime
from pandas.tseries.offsets import Day, MonthEnd
now = datetime(2011, 11, 17)
now + 3 * Day()
```

```python
# 如ME会前滚到第一个符合的日期
now + MonthEnd()
```

```python
offset = MonthEnd()
offset.rollforward(now)  # 显式前滚
```

```python
offset.rollback(now)  # 显式后滚
```
