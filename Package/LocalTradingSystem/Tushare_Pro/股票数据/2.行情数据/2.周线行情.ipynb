## 未复权

```python
# 描述:获取A股周线行情,本接口每周最后一个交易日更新,如需要使用每天更新的周线数据,请使用日度更新的周线行情接口
```

```python
import tushare as ts
pro = ts.pro_api()
df = pro.weekly(ts_code='601899.SH',start_date='20260101',end_date='20260225',fields='ts_code,trade_date,open,high,low,close,vol,amount')
df.head()
```

### 输入

```python
# ts_code TS代码
# trade_date 交易日期(每周最后一个交易日期,YYYYMMDD)
# start_date 开始日期
# end_date 结束日期
```

### 输出

```python
# ts_code 股票代码
# trade_date 交易日期
# close 周收盘价
# open 周开盘价
# high 周最高价
# low 周最低价
# pre_close 上一周收盘价
# change 周涨跌额
# pct_chg 周涨跌(未复权,未100,如果是复权请用 通用行情接口 如需%单位请100)
# vol 周成交量
# amount 周成交额
```
