```python
import tushare as ts
pro = ts.pro_api()
df = pro.bak_basic(trade_date='20211012', fields='trade_date,ts_code,name,industry,pe')
df.head(20)
```

### 输入

```python
# trade_date 交易日期
# ts_code 股票代码
```

### 输出

```python
# trade_date 交易日期
# ts_code 股票代码
# name 股票名称
# industry 所属行业
# pe 市盈率(动)
# float_share 流通股本(亿)
# total_share 总股本(亿)
# total_assets 总资产(亿)
# liquid_assets 流动资产(亿)
# fixed_assets 固定资产(亿)
# pb 市净率
# eps 每股收益
# bvps 每股净资产
# list_date 上市日期
# undp 未分配利润
# per_undp 每股未分配利润
# rev_yoy 营业收入同比(%)
# profit_yoy 净利润同比(%)
# gpr 毛利率(%)
# npr 净利润率(%)
# holder_num 股东人数
```
