```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

tips.head()
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",  # 按性别着色
    style="time",  # 按时间控制点样式
    size="size",  # 按人数控制点大小
    sizes=(20, 200),  # 点大小的最小/最大范围
    palette="Set2",  # 颜色映射方案，如 "Set2", "deep", "colorblind"
    alpha=0.7,  # 透明度
    legend="full"  # 是否显示图例，"auto", "brief", "full", False
)
plt.show()
```

```python
fmri = sns.load_dataset("fmri")

fmri
```

```python
sns.lineplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    style="region",
    markers=True,  # 数据点加标记
    dashes=[(2,2),(1,1)],  # 自定义线型
    errorbar="sd",  # 置信区间类型，"sd"（标准差）, "ci"（置信区间）, "pi"（预测区间）, None
    estimator="mean"  # 聚合方式，默认 "mean"（均值），可选 "median", "sum"
)
plt.show()
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    size="size",
    style="time",
    col="time",  # 按时间分列子图
    row="day",  # 按天分行子图
    height=4,  # 子图高度
    aspect=1.2,  # 宽高比
    kind="scatter",  # 或 "line"
    palette="muted"
)
plt.show()
```

```python
penguins = sns.load_dataset("penguins")

sns.histplot(
    data=penguins,
    x="flipper_length_mm",
    hue="species",
    bins=20,
    binwidth=None,  # 柱子宽度,与bins二选一
    multiple="dodge",  # 分组时的堆叠方式："layer"（叠加）, "stack"（堆叠）, "dodge"（并排）, "fill"（归一化堆叠）
    kde=True,  # 叠加密度曲线
    stat="density",  # 	统计量类型："count"（计数）, "density"（密度）, "frequency", "probability"
    common_norm=True  # 所有分组一起归一化
)
plt.show()
```

```python
sns.kdeplot(
    data=penguins,
    x="flipper_length_mm",
    hue="species",
    multiple="fill",  # 与 histplot 相同："layer", "stack", "fill"
    bw_method=0.3,  # 带宽选择方法，"scott", "silverman" 或数值（越小曲线越尖）
    cut=2,  # 曲线延伸超出数据范围的量（以带宽为单位），默认 3
    fill=True,  # 是否填充曲线下的面积
    palette="Set1"
)
plt.show()
```

```python
sns.displot(
    data=penguins,
    x="flipper_length_mm",
    hue="species",
    col="species",
    kind="kde",
    height=4,
    aspect=1.2,
    facet_kws={'sharex': False}  # 不同子图共享x轴（默认True）
)
plt.show()
```

```python
sns.boxplot(
    data=tips,
    x="day",  # x 放分类变量
    y="total_bill",  # y 放数值变量
    hue="sex",
    order=['Thu','Fri','sat','Sun'],  # 控制 x 轴类别的显示顺序
    hue_order=['Male','Female'],  # 控制 hue 分组的显示顺序控制 hue 分组的显示顺序
    palette="Set3",
    width=0.7,  # 每个箱线图在 x 轴类别上所占的宽度比例
    whis=1.5,  # 须线延伸到 IQR 的倍数，默认 1.5  IQR=Q3-Q1  [Q1 - whis*IQR, Q3 + whis*IQR]
    showfliers=True,  # 是否显示离群点，默认 True
    dodge=True  # hue 分组是否错开位置（避免重叠），默认 True
)
plt.show()
```

```python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    split=True,  # 布尔值，True 时将 hue 分组分置小提琴两侧（便于对比）
    inner="quartile",  # 内部标记类型："box"（箱线）, "quartile"（四分位线）, "point", "stick", None
    density_norm='area',
    palette="Set2",
    width=0.9,  # 小提琴宽度（比例）
    dodge=False
)
plt.show()
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    estimator="median",  # 统计函数，默认 "mean"（均值），可选 "median", "sum"
    errorbar="sd",  # 误差棒类型："sd", "ci", "pi", None
    palette="Blues_d",
    width=0.6,  # 条形宽度（比例）
    capsize=0.1
)
plt.show()
```

```python
sns.countplot(
    data=tips,
    x="day",
    hue="sex",
    order=['Thu','Fri','Sat','Sun'],
    palette='pastel',
    width=0.7,
    dodge=True
)
plt.show()
```

```python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    col="time",
    order=['Thu','Fri','Sat','Sun'],
    kind="box",  # "strip", "swarm", "box", "violin", "bar", "point", "count"
    height=4,
    aspect=0.8,
    palette="Set2"
)
plt.show()
```

```python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    order=2,  # 多项式回归的次数，默认 1（线性）
    ci=95,  # 置信区间，如 ci=95，None则不画
    scatter=True,  # 是否画散点，默认True
    fit_reg=True,  # 是否画回归线，默认True
    scatter_kws={'alpha':0.4, 's':30},  # 控制散点样式的额外参数（如 alpha,s）
    line_kws={'color':'red', 'lw':2},  # 控制回归线样式的额外参数（如 color, lw）
    color='blue',  # 统一点和线的颜色
    logx=False,  # 是否对 x 取对数再回归，默认False
    x_bins=10  # 将连续型的 x 轴变量分割成多个区间（分箱），然后对每个区间内的 y 值进行聚合统计（默认计算平均值），再绘制这些聚合后的点并拟合回归线。
)
plt.show()
```

```python
sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    height=4,
    aspect=0.9,
    order=1,
    ci=95,
    scatter_kws={'alpha':0.5},
    line_kws={'color':'red'}
)
plt.show()
```

```python
corr = tips.select_dtypes(include=['float64', 'int64']).corr()

corr
```

```python
sns.heatmap(
    data=corr,
    annot=True,  # 是否在每个格子中显示数值，默认False
    fmt=".2f",  # 数值格式，如 ".2f"（两位小数）, ".1f", "d"（整数）
    cmap="coolwarm",  # 颜色映射方案，如 "coolwarm", "RdYlGn", "viridis"
    center=0,  # 颜色居中值
    square=True,  # 是否让格子呈正方形
    linewidths=1,  # 格子边框宽度
    linecolor='skyblue',  # 格子边框颜色
    cbar=True,  # 是否显示颜色条，默认True
    cbar_kws={'shrink':0.8, 'label':'Correlation'},  # 调整颜色条样式
    annot_kws={'size':10, 'color':'black'},  # 调整注释字体样式
    xticklabels=True,  # 是否显示轴标签
    yticklabels=True  # 是否显示轴标签
)
plt.show()
```

```python
sns.pairplot(
    data=penguins,
    hue="species",
    palette="Set2",
    diag_kind="kde",  # 对角线图的类型："auto", "kde", "hist", None
    kind="scatter",  # 非对角线图类型："scatter"（默认）或 "reg"（带回归线）
    corner=False,  # 是否只画下三角矩阵（节省空间），默认False
    height=3,
    aspect=1.2,
    dropna=True,  # 是否忽略 NaN 行
    plot_kws={'alpha':0.5, 's':20},  # 	散点图的额外参数
    diag_kws={'fill':True, 'alpha':0.6},  # 对角线图的额外参数
    markers="+"  # 散点标记样式，如 "o", "+", "."
)
plt.show()
```
