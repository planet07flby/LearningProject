```python
# ts_code 股票代码
# ann_date 公告日期
# name 姓名
# age 年龄
# edu 学历
# title 职务
# gender 性别
# national 国籍
# begin_date 任职开始日期
# end_date 任职结束日期
# resume 个人简历
```
